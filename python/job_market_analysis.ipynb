```python
text_columns = ["title", "company", "location", "experience_level"]

for col in text_columns:
    vacancies[col] = vacancies[col].str.strip()
```

```python
# we are changing the datatype object to datetime of column "created_at" , "published_at".
vacancies["created_at"] = pd.to_datetime(vacancies["created_at"], errors="coerce")
vacancies["published_at"] = pd.to_datetime(vacancies["published_at"], errors="coerce")
```
